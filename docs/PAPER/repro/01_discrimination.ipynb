### E1.1 — 8-class LORO discrimination exceeds 0.125 chance at every ROI (both CVD)

```python
j = U.load_json(U.DEC / "model_comparison_validation/results/cvd_cross_decoding/cvd_cross_decoding_procrustes.json")
res = j["results"]
allpass = True
for roi in ["V1", "V2", "V3", "V4"]:
    ch = res[roi]["chance"]
    for s in ["sub-08", "sub-09"]:
        acc = res[roi]["cvd_cross"][s]
        ok = U.check(f"E1.1 {roi} {s} acc>chance", acc, ch, mode="ge")
        allpass = allpass and ok
print("all ROI/subj above chance:", allpass)
```

*Target:* both CVD exceed exact-accuracy chance 1/8=0.125 at V1–hV4.

### E1.2 — cross-subject generalization Mann–Whitney p=0.668 (LDA, all ROIs pooled)

```python
j = U.load_json(U.DEC / "results/loro/srm/validation/cross_subject_generalization.json")
p = j["LDA"]["difference"]["p_value"]
U.check("E1.2 MannWhitney p", p, 0.668, ndigits=3)
print("hc_to_hc mean=%.3f  hc_to_cvd mean=%.3f" % (j["LDA"]["hc_to_hc"]["mean"], j["LDA"]["hc_to_cvd"]["mean"]))
```

*Target:* p=0.668 (21 HC-to-HC vs 14 HC-to-CVD pairs).
