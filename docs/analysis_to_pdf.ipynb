# ColorBlind Analysis Documentation Generator

This notebook converts our analysis documentation into a PDF format for easier review and sharing. It includes:
1. Analysis timeline
2. Technical details per timestamp
3. Progress tracking
4. Next steps

```python
# Required libraries
import pandas as pd
import numpy as np
import markdown
import pdfkit
import os
from datetime import datetime

# Create docs directory if it doesn't exist
os.makedirs('analysis_output', exist_ok=True)
```

```python
# Read the markdown files
with open('../docs/analysis_log.md', 'r') as f:
    log_content = f.read()
    
with open('../docs/analysis_summary.md', 'r') as f:
    summary_content = f.read()

# Convert to HTML
html_content = f"""
<html>
<head>
<style>
body {{ font-family: Arial, sans-serif; }}
h1 {{ color: #2c3e50; }}
h2 {{ color: #34495e; }}
code {{ background-color: #f7f9fa; padding: 2px 4px; }}
</style>
</head>
<body>
{markdown.markdown(log_content)}
<hr>
{markdown.markdown(summary_content)}
</body>
</html>
"""

# Save HTML
with open('analysis_output/analysis_doc.html', 'w') as f:
    f.write(html_content)

# Convert to PDF
pdfkit.from_string(html_content, 'analysis_output/analysis_doc.pdf')
```

## Generated Documentation

The PDF has been generated with the following structure:
1. Analysis Log (Chronological timeline with timestamps)
2. Technical Summary (Detailed implementation notes)
3. Next Steps and Future Work

You can find the generated files in the `analysis_output` directory:
- analysis_doc.html: HTML version with formatting
- analysis_doc.pdf: PDF version for sharing
